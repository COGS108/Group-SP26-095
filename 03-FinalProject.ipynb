## Ethics

As given in class ethics is defined as “Moral principles that govern a person’s behaviour or the conducting of an activity”, we used publicly available data from the NCES Digest of Education Statistics. All data is aggregated by major and year, and it contains no personal information like names, student IDs, or addresses. This means we cannot identify any individual person, so privacy is not a concern.
Our classification of majors into "high AI exposure" (Computer Science, Engineering, Mathematics, Physics) and "low AI exposure" (History, Art, Philosophy, Political Science) is based on prior literature (Gao & Wang, 2023). This classification might not be perfect, but we were transparent about our method so readers can understand our reasoning.
We did not manipulate any data. We performed basic cleaning (renaming majors, removing missing values), but we never changed the actual unemployment numbers. Our analysis is reproducible, as if someone downloads the same raw data and runs our code, they should get the same results.

## Discusison and Conclusion

Previous research (Hajkowicz et al., 2023; Thayyib et al., 2023) has shown that AI adoption varies across fields, but few studies have examined post-2017 unemployment trends by major. Our study addresses this gap. We started with a hypothesis: after 2017 (when modern AI like Transformers emerged), high-AI-exposure majors would see their unemployment rates increase more than low-AI-exposure majors. Our results did not support this hypothesis.
Individual major analysis: 

For high-AI majors (Computer Science, Engineering, Mathematics, Physics), none of them showed a statistically significant post-2017 change in unemployment. All p-values were above 0.05. Their model fits were also weak (R-squared values between 0.100 and 0.376), meaning our interrupted time series model didn't explain much of what was happening for these majors.

For low-AI majors, the results were different. Political Science showed a statistically significant post-2017 increase (p = 0.003). Art and History were close to significant (p = 0.053 for both). Philosophy was not significant (p = 0.687). Model fits were much stronger for these majors, especially for Political Science (R-squared = 0.862) and History (R-squared = 0.740).

 When we grouped majors together, the high-AI group showed no significant post-2017 change (p = 0.319) and a poor model fit (R-squared = 0.172). The low-AI group showed a significant post-2017 increase (p = 0.042) and a much better model fit (R-squared = 0.679).
 
In summary, low-AI majors saw their unemployment go up after 2017, but high-AI majors did not.

Interpretation

We try to figure out why didn't AI-exposed majors see higher unemployment？ One possible explanation is that AI is augmenting technical work rather than replacing it — tools like Copilot help programmers but don't make them obsolete.
In additon, The COVID-19  happened right after 2017, and it likely had a major impact on unemployment, especially in 2020-2021. The sharp increase we see for low-AI majors in 2021 might be more about the pandemic than about AI. We cannot fully separate these two effects with our data.

Limitations

We need to be honest about the problems in our study.
First, we only have 10 years of data. This is a short time series, which gives us low statistical power. 
Second, the COVID-19 pandemic is a major confounder. We cannot separate COVID effects from AI effects. 
Third, our classification of "high AI exposure" and "low AI exposure" came from a single paper (Gao & Wang, 2023). Different papers might group majors differently. If we used a different classification, our results might change.

Future Work

If we continued this research, here is what we would do differently:
First, we would use more years of data. That would give us more statistical power and help us see real trends more clearly.
Second, we would try to control for the pandemic — for example, by adding a COVID variable or comparing across states with different policies.

Final Remarks

Even with these limitations, our study shows that after 2017, high-AI-exposure majors did not see a significant change in unemployment, while low-AI-exposure majors (especially Political Science) did. This challenges the common opinion that AI primarily threatens technical jobs. Our findings suggest that we should pay attention to a wider range of fields when thinking about AI's impact on the workforce.
