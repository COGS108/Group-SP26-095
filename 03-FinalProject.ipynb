## Abstract
Our study investigates whether the post-2017 rise of Transformer-based AI differently affected unemployment rates among college majors with high versus low AI exposure. We wanted to investigate this topic because we noticed that a lot of graduate students concern about employment prospects and decisions. As college students who are about to graduate, we think this is a significant topic to explore. 
By using the NCES Digest of Education Statistics (2012–2021), we compared the unemployment rates for high-AI-exposure majors like Computer Science, Engineering, Mathematics, and Physics and low-AI-exposure majors such as History, Art, Philosophy, and Political Science. We used difference-in-differences and interrupted time series models, setting 2017 as the year marking modern AI's emergence.
We hypothesized that high-AI-exposure majors would face larger unemployment rate increases, however, the results showed the opposite pattern. The fixed-effects difference model revealed that the post-2017 slope change for high-AI-exposure majors was 0.82 percentage points per year lower than for low-AI-exposure majors (p = 0.0096). Our category-level interrupted time series analysis showed low-AI-exposure majors experienced a significant increase in unemployment rate after 2017 (coefficient = 0.9540, p = 0.042), while high-AI-exposure majors showed no significant change. Among the low-AI exposure subjects, Political Science drove much of this effect (p = 0.003).
Individual major models for AI-exposed fields did not fit that well (R-squared values 0.100–0.376), suggesting unemployment rate in these fields is influenced by factors beyond a simple AI exposure trend. There are limitations in our study. First, we used only ten years of data. Second, we did not take into account the confounding effects from COVID-19 after 2017. In addition, we encountered observational survey data, and inability to prove causation.
Our findings suggest that unemployment rate has not been disturbed by high AI exposure. Technical majors maintained stable employment outcomes, while some non-technical majors saw relative impacts after 2017.

## Background and Prior Work

As many college students approach graduation, concerns about employment prospects and decisions regarding graduate education have become increasingly significant. The rapid development of artificial intelligence (AI), particularly following the introduction of the Transformer architecture in 2017, has substantially reshaped both scientific research and labor market demands. Previous research has documented the uneven adoption of AI across different academic fields. Hajkowicz et al. (2023) conducted a bibliometric analysis covering 333 research fields from 1960 to 2021 and found that AI adoption varies significantly across the physical sciences, natural sciences, life sciences, social sciences, and the arts and humanities. Thayyib et al. (2023) found that key AI technologies such as machine learning, deep learning, and natural language processing remain concentrated in specific domain clusters, while other fields such as engineering and biomedical sciences show relatively limited AI application. Additionally, Gao and Wang (2023) developed a framework to quantify both the direct use and potential benefits of AI across scientific disciplines, suggesting that AI may exacerbate existing inequalities across fields. While these studies provide valuable insights into the differential impact of AI across disciplines, relatively little research has examined how majors with different levels of AI exposure have experienced changes in employment outcomes and graduate school enrollment trends. This study aims to address this gap by analyzing post-2017 trends among UC San Diego students using data on employment outcomes, industries, graduation years, and graduate school enrollment across different majors. The goal is to investigate whether AI exposure has become a significant factor shaping undergraduate career and educational trajectories.

### NCES Field-of-Study Employment Outcomes Dataset


This dataset comes from NCES Digest of Education Statistics Table 505.10, which reports employment outcomes for 25- to 29-year-old bachelor’s degree holders by field of study. The original tables include metrics such as number of degree holders, percentage distribution, unemployment rate, median annual earnings, and the percentage of all 25- to 29-year-olds with a degree in each field. The NCES tables report standard errors in parentheses, which means many values are estimates based on survey data rather than exact population values. The data are based on American Community Survey 1-Year PUMS data. 

The main metric for this project is unemployment rate, measured as a percentage of the civilian labor force within each field of study. For example, an unemployment rate of 4.5 means that 4.5% of people in that field’s civilian labor force were unemployed. The unit is percentage points, so a change from 3.0 to 5.0 is a 2 percentage point increase. Other variables in the raw tables, such as number of degree holders and median annual earnings, are useful for context but are not the main dependent variable in our difference-in-differences model. NCES reports median earnings for full-time, year-round workers, while the number of degree holders is reported in thousands. 

For this project, we will extract unemployment rates for eight selected fields from 2012 to 2021. The high-AI-exposure group includes Computer Science, Engineering, Mathematics, and Physics. The low-AI-exposure group includes History, Art, Philosophy, and Political Science. After cleaning, the dataset will be tidy because each row will represent one major in one year, with separate columns for major, year, unemployment_rate, ai_exposure_group, post_2017, and did_interaction (It equals 1 only when both conditions are true: 1) The major is in the high-AI-exposure group. 2) The year is after 2017).

There are several concerns with this dataset. First, the raw NCES tables are not already in a tidy panel format, so the relevant rows and years must be extracted manually. Second, some field names do not perfectly match our simplified major names. For example, Computer Science appears as “Computer and information sciences,” Art appears as “Arts, fine and commercial,” and Physics may need to be represented by the broader “Physical sciences” category. Third, the data are observational, so the project cannot prove that AI exposure caused changes in unemployment rates. Finally, the analysis may be affected by other labor-market shocks, especially COVID-19, and by the small number of selected majors.

```python
# Specify Year
file_years = {
    'nces_505_10_d13.xls': 2012, 
    'nces_505_10_d14.xls': 2013, 
    'nces_505_10_d15.xls': 2014, 
    'nces_505_10_d16.xls': 2015,
    'nces_505_10_d17.xls': 2016,
    'nces_505_10_d18.xls': 2017,
    'nces_505_10_d19.xls': 2018,
    'nces_505_10_d20.xls': 2019,
    'nces_505_10_d21.xls': 2020,
    'nces_505_10_d22.xlsx': 2021,
}

# Define a major map
major_map = {
    'Computer and information sciences': 'Computer Science', 
    'Computer and information systems': 'Computer Science',
    'General engineering': 'Engineering',
    'Mathematics': 'Mathematics',
    'Physical sciences': 'Physics',
    'History': 'History',
    'Fine arts': 'Art',
    'Philosophy and religious studies': 'Philosophy',
    'Political science and government': 'Political Science',
}

# High AI Exposure
High_AI = ['Computer Science', 'Physics', 'Engineering', 'Mathematics']
```

```python
# Function for cleaning the files
def clean_file(file, year):
    path = 'data/00-raw/' + file
    df = pd.read_excel(path, header = None)
    df = pd.DataFrame({
        'major_raw': df.iloc[:, 1],
        'unemployment_rate': df.iloc[:, -9]
    })
    df = df[['major_raw', 'unemployment_rate']].copy()
    
    df['major_raw'] = df['major_raw'].str.replace(r'\.+', '', regex=True).str.replace(r'\s+', ' ', regex=True).astype(str).str.strip()

    df['unemployment_rate'] = pd.to_numeric(
        df['unemployment_rate'],
        errors='coerce'
    )
    
    df = df[df['major_raw'].isin(major_map.keys())].copy()

    df['major'] = df['major_raw'].map(major_map)
    df['year'] = year

    df = df[['major', 'year', 'unemployment_rate']]

    # drop any null values
    df = df.dropna()

    return df
```

```python
parts = []

# loop through each year
for file, year in file_years.items():
    template = clean_file(file, year)
    
    #store intermediate data
    template.to_csv(
        'data/01-interim/nces_' + str(year) + '_cleaned.csv',
        index=False
    )
    
    # add to the list
    parts.append(template)
    
# combine into pivot_long format
cleaned_df = pd.concat(parts, ignore_index=True)
```

```python
# add variable
cleaned_df['ai_exposure_group'] = cleaned_df['major'].isin(High_AI).astype(int)

cleaned_df['post_2017'] = (cleaned_df['year']>=2018).astype(int)

cleaned_df['did_interaction'] = cleaned_df['ai_exposure_group'] * cleaned_df['post_2017']
```

## Individual Major Interrupted Time Series Model
The interrupted time series models here estimate whether each individual major experienced a statistically meaningful change after 2017. Since each major only has ten years of observations, these models should be interpreted carefully. With a small sample size, the model may have a relatively low statistical power, and individual year fluctuations and outliers can strongly affect the prediction. 

For the AI-exposed majors, none of the individual major models show a statistically significant post-2017 slope change. Computer Science has a post coefficient of -0.1826 with a p-value of 0.623. Engineering has a post coefficient of 0.2425 with a p-value of 0.546. Mathematics has a post coefficient of -0.0369 with a p-value of 0.929. Physics has a larger positive post coefficient of 0.5118, but its p-value is 0.147. Therefore, none of the majors shows a significant change.

The model fit for the AI-exposed majors is also generally weak. Computer Science has an R-squared of 0.100, Engineering has an R-squared of 0.108, Mathematics has an R-squared of 0.202, and Physics has an R-squared of 0.376. This suggests that the Interrupted Time Series Model explains only a limited amount of variation in unemployment rates for these majors.



For the non-AI-exposed majors, the models shows greater statistical significance, especially for political science. Art has a post coefficient of 1.2306 with a p-value of 0.053, which shows a moderate level of significance but quite enough. History has a post coefficient of 0.4275 with a p-value of 0.053, also close to significance. Philosophy has a post coefficient of 0.3479 with a p-value of 0.687, showing no significant change. Political Science has a post coefficient of 1.8101 with a p-value of 0.003, which is statistically significant.

The model fit is also stronger for some non-AI-exposed majors. Art has an R-squared of 0.553, History has an R-squared of 0.740, Philosophy has an R-squared of 0.239, and Political Science has an R-squared of 0.862. Political Science has the strongest model fit, meaning that the interrupted time series structure explains a large portion of its unemployment-rate variation. 

```python
# interrupted time series analysis for each major
ai_exposed_models = {}
for major in ai_exposed['major'].unique():
    model = smf.ols("unemployment_rate ~ time + intervention + post", data=ai_exposed[ai_exposed['major'] == major]).fit()
    ai_exposed_models[major] = model
    print("\n" + "=" * 80)
    print(f"Major: {major}")
    print("=" * 80)
    print(model.summary())
not_ai_exposed_models = {}
for major in not_ai_exposed['major'].unique():
    model = smf.ols("unemployment_rate ~ time + intervention + post", data=not_ai_exposed[not_ai_exposed['major'] == major]).fit()
    not_ai_exposed_models[major] = model
    print("\n" + "=" * 80)
    print(f"Major: {major}")
    print("=" * 80)
    print(model.summary())
```

```python
def plot_interrupted_time_series(data, models, group_name):
    x_col = "year"

    for major in data["major"].unique():
        data_major = data[data["major"] == major].copy()
        data_major = data_major.sort_values(x_col)

        model = models[major]

        data_major["predicted"] = model.predict(data_major)

        counterfactual_data = data_major.copy()
        counterfactual_data["intervention"] = 0
        counterfactual_data["post"] = 0

        data_major["counterfactual"] = model.predict(counterfactual_data)

        post_data = data_major[data_major["intervention"] == 1]

        intervention_x = post_data[x_col].min()

        plt.figure(figsize=(10, 6))

        plt.scatter(
            data_major[x_col],
            data_major["unemployment_rate"],
            label="Observed unemployment rate"
        )

        plt.plot(
            data_major[x_col],
            data_major["unemployment_rate"],
            marker="o",
            alpha=0.5,
            label="Observed trend"
        )

        plt.plot(
            post_data[x_col],
            post_data["predicted"],
            label="Post-intervention fitted trend"
        )

        plt.plot(
            data_major[x_col],
            data_major["counterfactual"],
            linestyle="--",
            label="Counterfactual trend"
        )

        if intervention_x is not None:
            plt.axvline(
                x=intervention_x,
                linestyle=":",
                label="Intervention year"
            )

        plt.xlabel("Year")
        plt.ylabel("Unemployment Rate")
        plt.title(f"Interrupted Time Series: {major} ({group_name})")
        plt.legend()
        plt.grid(True)
        plt.show()

plot_interrupted_time_series(
    ai_exposed,
    ai_exposed_models,
    "AI-exposed majors"
)

plot_interrupted_time_series(
    not_ai_exposed,
    not_ai_exposed_models,
    "Non-AI-exposed majors"
)
```

## Summary of Individual Major Scale

At the individual-major scale, the results do not show strong evidence that AI-exposed majors experienced significant unemployment increases after 2017. None of the AI-exposed majors have statistically significant post coefficients, and their model fits are generally weak.

In contrast, the non-AI-exposed majors show stronger evidence of post-2017 changes. Political Science has a statistically significant post-intervention slope increase, while Art and History have post-intervention slope changes that are close to significance. This suggests that the unemployment rate increase after 2017  is more visible among some non-AI-exposed majors than among AI-exposed majors. 

This shows that the expectation that AI-exposed majors would experience larger unemployment increases after the rise of modern AI is false. Instead, the individual-major scale analysis shows that unemployment rates after 2017 may have changed more strongly for certain non-AI-exposed majors.

## Interrupted Time Series Plot
For AI-exposed majors, the observed data fluctuate substantially from year to year. The fitted post-intervention trend slopes slightly upward, but the regression model shows that this post-intervention slope change is not statistically significant. The counterfactual trend slopes downward, but the observed points do not consistently follow either the counterfactual or the fitted post-intervention line very closely.

This supports the conclusion that the interrupted time series model does not fit the AI-exposed category very well. The low R-squared value confirms this visually. The AI-exposed group does not show a strong or statistically reliable post-2017 shift in unemployment trend.

The interrupted time series plot for non-AI-exposed majors shows a clearer pattern.This visual difference between the downward counterfactual line and the upward post-intervention fitted line suggests that the unemployment trend for non-AI-exposed majors changed direction after 2017. The regression result supports this interpretation because the post coefficient is positive and statistically significant. Therefore, the graph and the model together suggest that non-AI-exposed majors experienced a meaningful upward shift in unemployment trend after the intervention period.

However, the graph also shows that the large increase in 2021 contributes heavily to this pattern. Therefore, the result should be interpreted carefully. The post-2017 increase may reflect a combination of AI-related labor market changes and other external factors, especially the post pandemic labor market.

```python
def plot_group_its(mean_data, model, group_name):
    mean_data = mean_data.sort_values("year").copy()

    mean_data["predicted"] = model.predict(mean_data)

    counterfactual_data = mean_data.copy()
    counterfactual_data["intervention"] = 0
    counterfactual_data["post"] = 0

    mean_data["counterfactual"] = model.predict(counterfactual_data)

    post_data = mean_data[mean_data["intervention"] == 1]

    intervention_year = post_data["year"].min()

    plt.figure(figsize=(10, 6))

    plt.scatter(
        mean_data["year"],
        mean_data["unemployment_rate"],
        label="Observed unemployment rate"
    )

    plt.plot(
        mean_data["year"],
        mean_data["unemployment_rate"],
        marker="o",
        label="Observed trend"
    )

    plt.plot(
        post_data["year"],
        post_data["predicted"],
        label="Post-intervention fitted trend"
    )

    plt.plot(
        mean_data["year"],
        mean_data["counterfactual"],
        linestyle="--",
        label="Counterfactual trend"
    )

    plt.axvline(
        x=intervention_year,
        linestyle=":",
        label="Intervention year"
    )

    plt.xlabel("Year")
    plt.ylabel("Mean Unemployment Rate")
    plt.title(f"Interrupted Time Series: {group_name}")
    plt.legend()
    plt.grid(True)
    plt.show()
    
plot_group_its(
    mean_ai_exposed,
    mean_ai_exposed_model,
    "AI-Exposed Majors"
)
plot_group_its(
    mean_not_ai_exposed,
    mean_not_ai_exposed_model,
    "Non-AI-Exposed Majors"
)
```

## Final Comparison: How the Change Differs Between AI-Exposed and Non-AI-Exposed Majors

The fixed-effects difference model provides the most direct test of whether the post-2017 change differs between AI-exposed and non-AI-exposed majors. The model has an R-squared of 0.483 and an adjusted R-squared of 0.382, meaning it explains a moderate amount of variation in unemployment rates. The overall model is statistically significant, with a Prob(F-statistic) of 9.80e-06. This suggests that the model as a whole explains unemployment variation better than a model with no predictors.

The key coefficient is post:ai_exposure_group, which measures whether the post-2017 slope change differs between AI-exposed and non-AI-exposed majors. This coefficient is -0.8203 with a p-value of 0.0096, meaning that the difference in post-2017 slope change between the two groups is statistically significant. 

The negative coefficient means that the post-2017 slope change for AI-exposed majors is about 0.82 percentage points per year lower than the post-2017 slope change for non-AI-exposed majors. In other words, non-AI-exposed majors experienced a stronger upward shift in unemployment trend after 2017, while AI-exposed majors were more stable.

Overall, the results support the alternative hypothesis that unemployment trends changed differently after 2017 for AI-exposed and non-AI-exposed majors. However, the direction of the result is different from what might be expected if AI exposure directly increased unemployment risk.

## Limitations
First, the dataset contains only ten years of data and only eight selected majors. This makes the individual-major models sensitive to year-to-year variation. Second, the post-2017 period overlaps with the COVID-19 pandemic, which likely affected unemployment rates across many fields. Third, AI exposure is measured as a broad category rather than as a direct measure of AI adoption within each occupation or industry. Fourth, the model shows association, not causation. Therefore, the results should be interpreted as evidence that unemployment trends differ by AI-exposure category, not as proof that AI caused the observed changes.

## Discusison and Conclusion

Previous research (Hajkowicz et al., 2023; Thayyib et al., 2023) has shown that AI adoption varies across fields, but few studies have examined post-2017 unemployment trends by major. Our study addresses this gap. We started with a hypothesis: after 2017 (when modern AI like Transformers emerged), high-AI-exposure majors would see their unemployment rates increase more than low-AI-exposure majors. Our results did not support this hypothesis.
Individual major analysis: 

For high-AI majors (Computer Science, Engineering, Mathematics, Physics), none of them showed a statistically significant post-2017 change in unemployment. All p-values were above 0.05. Their model fits were also weak (R-squared values between 0.100 and 0.376), meaning our interrupted time series model didn't explain much of what was happening for these majors.

For low-AI majors, the results were different. Political Science showed a statistically significant post-2017 increase (p = 0.003). Art and History were close to significant (p = 0.053 for both).  When we grouped majors together, the high-AI group showed no significant post-2017 change (p = 0.319) and a poor model fit (R-squared = 0.172). The low-AI group showed a significant post-2017 increase (p = 0.042) and a much better model fit (R-squared = 0.679).
 
In summary, low-AI majors saw their unemployment go up after 2017, but high-AI majors did not.

Limitations

We need to be honest about the problems in our study.
There are limitations. We only engaged with data between 10 years. Most importantly, the COVID-19 pandemic hit right after 2017, making it hard to tell AI effects from pandemic effects on the unemployment rate. In addition, the data come from surveys, not exact counts, which might encounter more biases. Another thing is that our classification of "high AI exposure" and "low AI exposure" came from a single paper (Gao & Wang, 2023). Different papers might group majors differently. If we used a different classification, our results might change. Furthermore, grouping majors into only "high" and "low" AI exposure rate oversimplifies reality. Therefore, we cannot prove AI caused these differences. We can only say that the patterns exist.
