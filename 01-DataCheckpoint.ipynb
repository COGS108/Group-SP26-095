### NCES Field-of-Study Employment Outcomes Dataset


This dataset comes from NCES Digest of Education Statistics Table 505.10, which reports employment outcomes for 25- to 29-year-old bachelor’s degree holders by field of study. The original tables include metrics such as number of degree holders, percentage distribution, unemployment rate, median annual earnings, and the percentage of all 25- to 29-year-olds with a degree in each field. The NCES tables report standard errors in parentheses, which means many values are estimates based on survey data rather than exact population values. The data are based on American Community Survey 1-Year PUMS data. 

The main metric for this project is unemployment rate, measured as a percentage of the civilian labor force within each field of study. For example, an unemployment rate of 4.5 means that 4.5% of people in that field’s civilian labor force were unemployed. The unit is percentage points, so a change from 3.0 to 5.0 is a 2 percentage point increase. Other variables in the raw tables, such as number of degree holders and median annual earnings, are useful for context but are not the main dependent variable in our difference-in-differences model. NCES reports median earnings for full-time, year-round workers, while the number of degree holders is reported in thousands. 

For this project, we will extract unemployment rates for eight selected fields from 2012 to 2022. The high-AI-exposure group includes Computer Science, Engineering, Mathematics, and Physics. The low-AI-exposure group includes History, Art, Philosophy, and Political Science. After cleaning, the dataset will be tidy because each row will represent one major in one year, with separate columns for major, year, unemployment_rate, ai_exposure_group, post_2017, and did_interaction.

There are several concerns with this dataset. First, the raw NCES tables are not already in a tidy panel format, so the relevant rows and years must be extracted manually. Second, some field names do not perfectly match our simplified major names. For example, Computer Science appears as “Computer and information sciences,” Art appears as “Arts, fine and commercial,” and Physics may need to be represented by the broader “Physical sciences” category. Third, the data are observational, so the project cannot prove that AI exposure caused changes in unemployment rates. Finally, the analysis may be affected by other labor-market shocks, especially COVID-19, and by the small number of selected majors.
