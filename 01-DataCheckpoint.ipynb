## Background and Prior Work

As many college students approach graduation, concerns about employment prospects and decisions regarding graduate education have become increasingly significant. The rapid development of artificial intelligence (AI), particularly following the introduction of the Transformer architecture in 2017, has substantially reshaped both scientific research and labor market demands. Previous research has documented the uneven adoption of AI across different academic fields. Hajkowicz et al. (2023) conducted a bibliometric analysis covering 333 research fields from 1960 to 2021 and found that AI adoption varies significantly across the physical sciences, natural sciences, life sciences, social sciences, and the arts and humanities. Thayyib et al. (2023) found that key AI technologies such as machine learning, deep learning, and natural language processing remain concentrated in specific domain clusters, while other fields such as engineering and biomedical sciences show relatively limited AI application. Additionally, Gao and Wang (2023) developed a framework to quantify both the direct use and potential benefits of AI across scientific disciplines, suggesting that AI may exacerbate existing inequalities across fields. While these studies provide valuable insights into the differential impact of AI across disciplines, relatively little research has examined how majors with different levels of AI exposure have experienced changes in employment outcomes and graduate school enrollment trends. This study aims to address this gap by analyzing post-2017 trends among UC San Diego students using data on employment outcomes, industries, graduation years, and graduate school enrollment across different majors. The goal is to investigate whether AI exposure has become a significant factor shaping undergraduate career and educational trajectories.

## Hypothesis


**Null Hypothesis:**
There is no difference-in-differences effect in unemployment rates between high-AI-exposure majors (Computer Science, Engineering, Mathematics, and Physics) and low-AI-exposure majors (History, Art, Philosophy, and Political Science) after 2017.

**Alternative Hypothesis:**
There is a significant difference-in-differences effect in unemployment rates between high-AI-exposure majors and low-AI-exposure majors after 2017.

We hypothesize that there is a significant difference-in-differences effect in unemployment rates between high-AI-exposure majors and low-AI-exposure majors after 2017 (the rise of Transformer). We think so because majors with higher exposure to AI are more likely to play significant roles in the rapid expansion of AI in various industries, therefore vastly impacting students’ future employment and unemployement.

### NCES Field-of-Study Employment Outcomes Dataset


This dataset comes from NCES Digest of Education Statistics Table 505.10, which reports employment outcomes for 25- to 29-year-old bachelor’s degree holders by field of study. The original tables include metrics such as number of degree holders, percentage distribution, unemployment rate, median annual earnings, and the percentage of all 25- to 29-year-olds with a degree in each field. The NCES tables report standard errors in parentheses, which means many values are estimates based on survey data rather than exact population values. The data are based on American Community Survey 1-Year PUMS data. 

The main metric for this project is unemployment rate, measured as a percentage of the civilian labor force within each field of study. For example, an unemployment rate of 4.5 means that 4.5% of people in that field’s civilian labor force were unemployed. The unit is percentage points, so a change from 3.0 to 5.0 is a 2 percentage point increase. Other variables in the raw tables, such as number of degree holders and median annual earnings, are useful for context but are not the main dependent variable in our difference-in-differences model. NCES reports median earnings for full-time, year-round workers, while the number of degree holders is reported in thousands. 

For this project, we will extract unemployment rates for eight selected fields from 2012 to 2021. The high-AI-exposure group includes Computer Science, Engineering, Mathematics, and Physics. The low-AI-exposure group includes History, Art, Philosophy, and Political Science. After cleaning, the dataset will be tidy because each row will represent one major in one year, with separate columns for major, year, unemployment_rate, ai_exposure_group, post_2017, and did_interaction (It equals 1 only when both conditions are true: 1) The major is in the high-AI-exposure group. 2) The year is after 2017).

There are several concerns with this dataset. First, the raw NCES tables are not already in a tidy panel format, so the relevant rows and years must be extracted manually. Second, some field names do not perfectly match our simplified major names. For example, Computer Science appears as “Computer and information sciences,” Art appears as “Arts, fine and commercial,” and Physics may need to be represented by the broader “Physical sciences” category. Third, the data are observational, so the project cannot prove that AI exposure caused changes in unemployment rates. Finally, the analysis may be affected by other labor-market shocks, especially COVID-19, and by the small number of selected majors.

```python
# import packages
import pandas as pd
```

```python
# Specify Year
file_years = {
    'nces_505_10_d13.xls': 2012, 
    'nces_505_10_d14.xls': 2013, 
    'nces_505_10_d15.xls': 2014, 
    'nces_505_10_d16.xls': 2015,
    'nces_505_10_d17.xls': 2016,
    'nces_505_10_d18.xls': 2017,
    'nces_505_10_d19.xls': 2018,
    'nces_505_10_d20.xls': 2019,
    'nces_505_10_d21.xls': 2020,
    'nces_505_10_d22.xlsx': 2021,
}

# Define a major map
major_map = {
    'Computer and information sciences': 'Computer Science', 
    'Computer and information systems': 'Computer Science',
    'General engineering': 'Engineering',
    'Mathematics': 'Mathematics',
    'Physical sciences': 'Physics',
    'History': 'History',
    'Fine arts': 'Art',
    'Philosophy and religious studies': 'Philosophy',
    'Political science and government': 'Political Science',
}

# High AI Exposure
High_AI = ['Computer Science', 'Physics', 'Engineering', 'Mathematics']
```

```python
# take a look
check = pd.read_excel('data/00-raw/nces_505_10_d17.xls', header=None)
check
```

```python
# Function for cleaning the files
def clean_file(file, year):
    path = 'data/00-raw/' + file
    df = pd.read_excel(path, header = None)
    df = pd.DataFrame({
        'major_raw': df.iloc[:, 1],
        'unemployment_rate': df.iloc[:, -9]
    })
    df = df[['major_raw', 'unemployment_rate']].copy()
    
    df['major_raw'] = df['major_raw'].str.replace(r'\.+', '', regex=True).str.replace(r'\s+', ' ', regex=True).astype(str).str.strip()

    df['unemployment_rate'] = pd.to_numeric(
        df['unemployment_rate'],
        errors='coerce'
    )
    
    df = df[df['major_raw'].isin(major_map.keys())].copy()

    df['major'] = df['major_raw'].map(major_map)
    df['year'] = year

    df = df[['major', 'year', 'unemployment_rate']]

    # drop any null values
    df = df.dropna()

    return df
```

```python
parts = []

# loop through each year
for file, year in file_years.items():
    template = clean_file(file, year)
    
    #store intermediate data
    template.to_csv(
        'data/01-interim/nces_' + str(year) + '_cleaned.csv',
        index=False
    )
    
    # add to the list
    parts.append(template)
    
# combine into pivot_long format
cleaned_df = pd.concat(parts, ignore_index=True)
```

```python
# add variable
cleaned_df['ai_exposure_group'] = cleaned_df['major'].isin(High_AI).astype(int)

cleaned_df['post_2017'] = (cleaned_df['year']>=2018).astype(int)

cleaned_df['did_interaction'] = cleaned_df['ai_exposure_group'] * cleaned_df['post_2017']
```
